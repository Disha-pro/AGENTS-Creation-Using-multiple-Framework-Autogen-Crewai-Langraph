"In the last notebook, we saw two AI agents talking to each other      



Until now, our AI agents could only communicate with each other.

But in a real application,

A customer asks a question.
A user uploads a document.
A manager gives an instruction.

The AI system needs someone to receive the user's request and coordinate the conversation.

That "someone" is the UserProxyAgent

```python
# instead of  Research Agent -  Writer Agent
# now we go for
# Human -userProxyAgent - Research Agent --Writer Agent --Final Answer --Human

# Now the conversation starts with the user instead of another AI.

# What is a UserProxyAgent?

# Simple definition: # A UserProxyAgent represents the human user.
#It receives the user's request, sends it to other agents, and returns the final response.

# Think of it as a bridge between the human and the AI agents.
```

```python
# Imagine a company.

# Customer -Receptionist - Developer - Tester - Receptionist - Customer
# The receptionist doesn't solve the problem.The receptionist coordinates the work.

# similarly --- User -- UserProxyAgent-- AI Agents -- UserProxyAgent--  User
# The receptionist is the proxy agent here
```

SEQUENTIAL CHAT

```python
strategy_config_list = {"config_list":[{"model":"gpt-4o-mini"}]}
```

```python
# First agent - gather the information basis the problem statement
strategy_creator = autogen.AssistantAgent(name="strategy_creator",
                                          llm_config=strategy_config_list,
                                          system_message=""" You are a logical marketing manager tasked with crafting a
                                          marketing strategy. Gether details on the product, target audience, and budget.
                                          Then identify the best marketing channels which will get the product maximum
                                          visibility.""",
                                         )
```

```python
content_config_list = {"config_list":[{"model":"gpt-4o-mini"}]}
```

```python
# 2nd Agent - create content basis the given input by first agent

content_creator = autogen.AssistantAgent(name="content_creator",
                                          llm_config=content_config_list,
                                          system_message=""" You are a Content Executor, a creative AI Agent.
                                          Your task is to craft engaging, channel-specific content that alings
                                           with the marketing strategy, ensuring it captures attention and
                                           resonates with the target audience.""",
                                         )
```

PROXY AGENT

```python
user_proxy = autogen.UserProxyAgent(
    name="User",
    human_input_mode="ALWAYS",
     is_termination_msg=lambda x:x.get("content", "") and x.get("content","").rstrip().endswith("TERMINATE"),
    code_execution_config=False,)

# is_termination_msg  HELPS IDENTIFIES WHEN THE CONVERSATION IS FINISHED OR ENDED

# x:x.get("content", "") -----FIRST CHECK IF CONTENT EXISTS - IS TEHRE ANY MESSAGE OR ITS AN EMPTY STRING
# .get("content","").rstrip().endswith("TERMINATE"  ---ITS CHECKING DOES MESSAFE END WITH TERMINATE

#   code_execution_config=False,)  - THIS TELLS PROXY DONT EXECUTE THE CODE AUTOMATICALLY  . ALWAYS INVOLVE HUMAN FOR FURTHER ACTION
```

```python
product_details = [
"""Water bottle for school childern under the age of 16 years. The marketing budget is 2000 USD""",
]
```

```python
writing_task = ["""Write an engaging content for the mentioned product and channel"""]
```

DEFINE THE SEQUENTIAL CHAT FLOW

```python
chat_results = autogen.initiate_chats(
    [
        {"sender": user_proxy,
         "recipient":strategy_creator,
         "message": product_details[0],
         "clear_history":True,
         "silent":False,
         "summary_method":"last_msg",},
        {"sender": user_proxy,
         "recipient":content_creator,
         "message": writing_task[0],
         "summary_method":"last_msg",},]
)
```
